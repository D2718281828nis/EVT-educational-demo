# 2. Масштабирование и зависимость

**Цели:** оценить зависимость приращений от лага и проверить, согласуется ли ряд с масштабным (фрактальным) поведением. Это диагностика, не доказательство фрактальности.

```python
from pathlib import Path
import os
import numpy as np
import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data/generated/fractal_extreme_series.csv").exists())
DATA = ROOT / "data/generated/fractal_extreme_series.csv"
SEED = 42
rng = np.random.default_rng(SEED)
df = pd.read_csv(DATA)
x = df["value"].to_numpy()
assert np.isfinite(x).all() and x.ndim == 1
```

```python
import matplotlib.pyplot as plt
lags=np.unique(np.logspace(0,2.6,35).astype(int))
q=(1,2,3)
structure={order: np.array([np.mean(np.abs(x[lag:]-x[:-lag])**order) for lag in lags]) for order in q}
slopes={order: np.polyfit(np.log(lags),np.log(vals),1)[0] for order,vals in structure.items()}
fig,ax=plt.subplots(figsize=(6,4))
for order,vals in structure.items(): ax.loglog(lags,vals,'o-',ms=3,label=f'q={order}, slope={slopes[order]:.2f}')
ax.set(xlabel='lag',ylabel='structure function',title='Scaling diagnostic'); ax.legend(); ax.grid(alpha=.2);
slopes
```

Разные наклоны для порядков $q$ мотивируют анализ неоднородности масштаба, но конечный ряд, тренд и выбросы могут создать тот же эффект.
